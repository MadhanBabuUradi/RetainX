```python
import numpy as np
import pandas as pd
import os
```

```python
df = pd.read_csv("../data/rawdata/employee_attrition_dataset_10000.csv")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:", df.isnull().sum())
print("\nData types:", df.dtypes)
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
retainx_df = df.sample(
    n=1000,
    random_state=42
).reset_index(drop=True)

print("RetainX Dataset Created!")
print("Shape:", retainx_df.shape)
```

```python
output_directory = "../data/processed"
os.makedirs(output_directory, exist_ok=True)
output_path="../data/processed/retainx_hr_1000.csv"
retainx_df.to_csv(output_path, index=False)
print("Dataset saved successfully!", output_path)
```

```python
# Attrition = "Yes": The employee has left the company (resigned, terminated, or retired).
# Attrition = "No": The employee is active and currently still working in the organization.
```

```python
# Employees Attrition rate :
# Employees who left
# ──────────────────── × 100
# Total employees

```

```python
totalemp=len(retainx_df)
attrition_count=(retainx_df["Attrition"]=="Yes").sum()
active_count=(retainx_df["Attrition"]=="No").sum()

attrition_rate=(attrition_count/totalemp)*100

print(f"Total employees : {totalemp}")
print(f"Active employees : {active_count}")
print(f"Attrition employees : {attrition_count}")
print(f"Overall Attrition rate : {attrition_rate:.2f}%")
```
