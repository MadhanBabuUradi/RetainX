### Salary Band

Employees are divided into Low, Medium, and High salary bands using income tertiles.

These bands are relative to the RetainX dataset and should not be interpreted as fixed market salary ranges.

```python
retainx_df["Salary_Band"] = pd.qcut(retainx_df["Monthly_Income"],
    q=3,
    labels=["Low", "Medium", "High"])

print(retainx_df["Salary_Band"].value_counts().sort_index())
```

```python
retainx_df["Overtime_Exposure"] = (retainx_df["Overtime"].map({"Yes": 1, "No": 0}))

print(retainx_df[["Overtime", "Overtime_Exposure"]].head(10))
```

```python
print(retainx_df["Overtime_Exposure"].value_counts(dropna=False))
```

```python
print("New Features:")
print(retainx_df[
        [
            "Career_Stagnation_Ratio",
            "Salary_Band",
            "Overtime_Exposure"
        ]
    ].head(10))

print("\nMissing values:")
print(retainx_df[
        [
            "Career_Stagnation_Ratio",
            "Salary_Band",
            "Overtime_Exposure"
        ]].isnull().sum())
```

```python
stagnation_summary = (retainx_df.groupby("Attrition")["Career_Stagnation_Ratio"].agg(["count", "mean", "median"]))

print(stagnation_summary.round(3))
```
